```python

# 1. Math Library
import math

print("Square Root:", math.sqrt(25))
print("Power:", math.pow(2, 3))
print("Factorial:", math.factorial(5))

# 2. NumPy Library
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print("\nNumPy Array:", arr)
print("Mean:", np.mean(arr))
print("Maximum:", np.max(arr))
print("Minimum:", np.min(arr))

# 3. SciPy Library
from scipy import stats

print("\nMode:", stats.mode(arr, keepdims=True).mode)
print("Z-score:", stats.zscore(arr))

# 4. Pandas Library
import pandas as pd

data = {
    'Name': ['Amit', 'Riya', 'Rahul', 'Sneha'],
    'Marks': [85, 90, 75, 95]
}

df = pd.DataFrame(data)

print("\nDataFrame:")
print(df)

print("\nStatistical Summary:")
print(df.describe())

# 5. Matplotlib Library
import matplotlib.pyplot as plt

plt.plot(df['Name'], df['Marks'], marker='o')
plt.title("Student Marks")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()

# 6. Seaborn Library
import seaborn as sns

sns.barplot(data=df, x='Name', y='Marks')
plt.title("Student Marks using Seaborn")
plt.show()
```
